```python
import numpy as np
import pandas as pd
```

```python
movies=pd.read_csv('tmdb_5000_movies.csv')
credits=pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies.head()
```

```python
credits.head(1)
```

```python
movies=movies.merge(credits,on='title')
```

```python
movies=movies[['movie_id','title','overview','genres','keywords','cast','crew']] 
```

```python
import ast #abstract syntax tree
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj): #literal_eval() is used to safely evaluate a string containing a Python literal or container display. It can be used to parse strings that represent Python data structures, such as lists, dictionaries, tuples, etc.
        L.append(i['name'])
    return L
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies['genres']=movies['genres'].apply(convert)
movies['keywords']=movies['keywords'].apply(convert)
```

```python
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

```python
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python

movies['cast'] = movies['cast'].apply(lambda x:x[0:3])
```

```python
movies.head()
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split()) #converts the overview into list of words as it was a string before
```

```python

# this function is used to remove spaces between the words in the list of genres,keywords,cast and crew because when we will do vectorization it will consider them as two different words and we want to consider them as one word
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python

new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
new.head()
```

```python
#converting the list of tags into string because when we will do vectorization it will consider them as two different words and we want to consider them as one word
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

```python
new['tags']=new['tags'].apply(lambda x: x.lower())
```

IMPORTING NLTK TO PERFORM STEMMING THAT CONVERTS WORDS INTO THEIR ROOT WORDS

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer #a word stemming algorithm that reduces words to their root/base form (called a "stem") by stripping suffixes.
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new['tags']=new['tags'].apply(stem)
```

TEXT VECTORIZATION USING BAG OF WORDS

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
vectors = cv.fit_transform(new['tags']).toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
  movie_index=new[new['title']==movie].index[0]
  distances = similarity[movie_index]
  movie_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
  for i in movie_list:
    print(new.iloc[i[0]].title)
```

```python
recommend('Batman Begins')
```
